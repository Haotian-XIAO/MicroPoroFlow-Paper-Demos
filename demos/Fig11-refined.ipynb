# Figure 11 (2/3): refined-mesh loading cases

**Symmetric loadings on a refined mesh**

This notebook recomputes, on a refined mesh ($l = 0.015$, refined to $r_0/10$ near the pores; MUMPS), the symmetric loading cases of Fig. 10 (equibiaxial stretching up to $\tilde{J} = 1.69$, gas pressure 0.5 kPa) for the nine $r_0$ of Fig. 10, and the case "equibiaxial stretching + 0.2 kPa" of Fig. 9 ($r_0 = 0.41$). For these loadings the homogenized permeability is isotropic, so only the probe $g_x$ is run, except for five cases that are run with both probes to check the isotropy. The results are used by `Fig11.ipynb` (colored symbols of Fig. 11(b)).

The last section compares the refined and the default mesh (Sec. 2.4); it needs the results of `Fig10.ipynb`.

The 24 runs are executed one after the other and take about 1.5 to 2 hours.

## Imports

```python
from pathlib import Path

import Runs_Microporoflow as rm
import Post_Microporoflow as pm

# Quieter output (messages only, results unchanged; see Quiet_Microporoflow.py):
# no gmsh information/progress lines, no ufl FutureWarning
import Quiet_Microporoflow as qm

qm.quiet_fe_output()
```

## Output directory

```python
# All results of this notebook are written to the subdirectory Fig11-refined/
# of the notebook directory (demos/).
RESULTS_DIR = Path("Fig11-refined").resolve()
RESULTS_DIR.mkdir(parents=True, exist_ok=True)

print("RESULTS_DIR =", qm.display_path(RESULTS_DIR))

# Default-mesh results of Fig10.ipynb (mesh-convergence check)
FIG10_DIR = Path("Fig10").resolve() / "Figure10_loading_comparison"
```

## Parameters

```python
do_run_refined      = 1  # finite-element runs on the refined mesh (24 runs)
do_mesh_convergence = 1  # post-processing only: refined vs default mesh (Sec. 2.4), needs Fig10.ipynb

refined_tag = "l0.015_local"
mesh = dict(l=0.015, local=True, solver="mumps")       # l = 0.015, refined to r0/10 near the pores, MUMPS
loading = dict(gamp=0.01, pg=0.5, Umax=0.3, beta=0.3)  # as in Fig10.ipynb


def case(mode, r0, probe, prefix="Fig10", pf_lst=None):
    return dict(mode=mode, r0=r0, probe=probe, prefix=prefix, pf_lst=pf_lst, **mesh, **loading)


cases_refined = (
    # equibiaxial stretching r0 = 0.07, probes gx and gy (also used for Sec. 2.4)
    [case("volumic", 0.07, probe) for probe in ("gx", "gy")]
    # equibiaxial stretching r0 = 0.1 to 0.2 and gas pressure r0 = 0.07, probes gx and gy (isotropy check)
    + [case("volumic", r0, probe) for r0 in (0.1, 0.15, 0.2) for probe in ("gx", "gy")]
    + [case("gas_pressure", 0.07, probe) for probe in ("gx", "gy")]
    # remaining r0, probe gx only
    + [case("volumic", r0, "gx") for r0 in (0.25, 0.3, 0.35, 0.41, 0.45)]
    + [case("gas_pressure", r0, "gx") for r0 in (0.1, 0.15, 0.2, 0.25, 0.3, 0.35, 0.41, 0.45)]
    # Fig. 9: equibiaxial stretching with a gas pressure of 0.1 kPa (end of step 1) and 0.2 kPa (end of step 2)
    + [case("volumic", 0.41, "gx", prefix="Fig9", pf_lst=(0.1, 0.2))]
)

convergence_r0 = (0.07, 0.10)  # Sec. 2.4

print(len(cases_refined), "refined-mesh runs")
print("material parameters:", rm.MATERIAL_PARAMETERS)
```

## Computing the refined-mesh loading cases

```python
if do_run_refined:
    rm.run_case_list(rm.run_refined_case, cases_refined, RESULTS_DIR / refined_tag)
```

## Mesh convergence (Sec. 2.4)

```python
if do_mesh_convergence:
    pm.require_files({
        "Fig10.ipynb": [
            f"{pm.fig10_basename(FIG10_DIR, 'volumic', r0)}-{probe}-qois.dat"
            for r0 in convergence_r0
            for probe in ("gx", "gy")
        ],
        "Fig11-refined.ipynb (this notebook, do_run_refined = 1)": [
            f"{pm.fig10_basename(RESULTS_DIR / refined_tag, 'volumic', r0)}-{probe}-qois.dat"
            for r0 in convergence_r0
            for probe in ("gx", "gy")
        ],
    })
    convergence_rows = pm.mesh_convergence_rows(RESULTS_DIR / refined_tag, FIG10_DIR, convergence_r0)
    pm.print_mesh_convergence(convergence_rows, RESULTS_DIR / "mesh_convergence_sec2.4.csv")
```
